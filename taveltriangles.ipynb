```python
df = pd.DataFrame({
    "Name": Name,
    "Stay": stay,
    "Location": location,
    "Discount": discount,
    "Price": Price,
    "Packages": Packages,
    "Page Number": pagenum})
```

```python
df["discount_price"] =( df["Price"].astype(str).str.findall(r"\d{1,3}(?:,\d{2})*,\d{3}").str[0].str.replace(",","").astype(int))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.boxplot(
    x='Locations_List',
    y='Discount_percentage',
    data=df
)
plt.xticks(rotation=45)
plt.title('Location vs Discount Price')
plt.show()
```

###   Insights:
- Munnar, Alleppey, and Thekkady show higher median discount percentages and wider variability, indicating more aggressive and varied discounting in these popular tourist locations.
- Kochi has a lower and tightly clustered discount percentage, suggesting more consistent and moderate discounts with fewer extreme offers.
- Locations such as Wayanad, Athirapally, Kovalam, Trivandrum, and Kumarakom exhibit almost fixed discount values, implying standardized or flat discount policies.
- The presence of outliers (e.g., in Kochi) indicates occasional promotional or special discount offers beyond the usual range.
